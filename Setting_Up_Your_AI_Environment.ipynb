```python
# 1.1 Lists
fruits = ["apple", "banana", "cherry"]
print("Fruits:", fruits)

numbers = [10, 20, 30, 40, 50]
print("Second element:", numbers[1])
print("Fourth element:", numbers[3])
```

```python

# 1.2 Tuples
measurements = (21.5, 22.8, 19.6, 20.0)
print("Tuple:", measurements)
print("First value:", measurements[0])
print("Last value:", measurements[-1])
```

```python

# 1.3 Dictionaries
book = {"title": "1984", "author": "George Orwell", "year": 1949}
print("Dictionary:", book)
print("Book title:", book["title"])
```

```python
# Task 2: Functions
def calculate_difference(num1, num2):
    return num1 - num2

def safe_divide(num1, num2):
    if num2 == 0:
        return "Cannot divide by zero"
    return num1 / num2

print("Difference:", calculate_difference(7, 2))
print("Division:", safe_divide(10, 4))
print("Division by zero:", safe_divide(10, 0))
```

```python
# 3.1 Loops
for i in range(3):
    print("Iteration:", i)

animals = ["cat", "dog", "bird"]
for animal in animals:
    print(animal)
```

```python
# 3.2 Conditionals
x = 10
if x > 5:
    print("x is greater than 5")

number = 4
if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```
